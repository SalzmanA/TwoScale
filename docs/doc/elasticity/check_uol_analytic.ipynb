# Analytic field for 3D plate test case


```python
import sympy as sp
from IPython.display import display, Math
from sympy import latex as spl_
import numpy as np
sp.init_printing()
def spl(expr):
    return(spl_(expr,mul_symbol='dot'))
```

```python
E,nu=sp.symbols('E nu',real=True)
lmbda,mu=sp.symbols('lambda mu',real=True)
x,y,z=sp.symbols('x y z',real=True)
L,B, H, f=sp.symbols('L B H f',real=True)
```

```python
lmbdaEN =(E*nu)/((1+nu)*(1.-2.*nu))
muEN =E/(2.*(1.+nu))
coefu=(nu+1)*(1-2*nu)*f/E
```

## Field expresion

```python
u=sp.Matrix([[x*x*(L/2-x/3)+2*nu*(y*y-z*z)],[-4*nu*x*y],[4*nu*x*z]])
display(Math(f' u = {spl(coefu)}{spl(u)}')) 
```

## Strain

Considering small displacements, the strain is given by:


```python
eps=sp.zeros(3)
for i in range(3):
    j=0
    for k in [x,y,z]:
        eps[i,j]=sp.diff(u[i],k)/2
        j+=1
for j in range(3):
    i=0
    for k in [x,y,z]:
        eps[i,j]+=sp.diff(u[j],k)/2
        i+=1
eps=sp.simplify(eps)
exp='\\frac{1}{2}.(u_{i,j}+u_{j,i})'
display(Math(f' eps = {exp}={spl(coefu)}{spl(eps)}')) 
```

## Stress

Considering Hook's law, constrains are given using Lamè coéficient for material properties:


```python
I=sp.eye(3)
tr=sp.trace(eps)
sig=sp.simplify(coefu*(2*mu*eps+lmbda*tr*I))
display(Math(f' sig=2.{spl(mu)}.eps +{spl(lmbda)}.I.trace(eps) = {spl(sig)}'))
```

Or if Lamè are replaced by engenring constant:

```python
sigEnu=sp.simplify(sig.subs( [(mu,muEN),(lmbda,lmbdaEN)] ))
display(Math(f' sig=2.{spl(mu)}.eps +{spl(lmbda)}.I.trace(eps) = {spl(sigEnu)}'))
```

## Loads

Equilibrium equations give then the following volumic load:

```python
div=sp.simplify(sp.diff(sigEnu[0,0],x)+sp.diff(sigEnu[1,1],y)+sp.diff(sigEnu[2,2],z))
fv=sp.factor(-div)
display(Math(f' fv=-div(sig) = {spl(fv)}'))
```

And on boundary ($\sigma.n$):

```python
fsyz=sp.factor(sp.Matrix([sigEnu[0,0],0.,0.]))
fsxy=sp.factor(sp.Matrix([0.,0.,sigEnu[2,2]]))
fsxz=sp.factor(sp.Matrix([0.,sigEnu[1,1],0.]))

display(Math(f' fs (x=L) = {spl(fsyz.subs([(x,L)]))} fs (x=0) = {spl(-fsyz.subs([(x,0)]))}'))
display(Math(f' fs (y=B) = {spl(fsxz.subs([(y,B)]))} fs (y=0) = {spl(-fsxz.subs([(y,0)]))}'))
display(Math(f' fs (z=H) = {spl(fsxy.subs([(z,H)]))} fs (z=0) = {spl(-fsxy.subs([(z,0)]))}'))
```
